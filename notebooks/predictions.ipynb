```python
def debug_preprocessing(train_data, test_data):
    """
    Simplified preprocessing to debug column alignment and data transformations.
    """
    print("\n--- Debugging Preprocessing ---")
    
    # Print the first few rows of the data
    print("Training data (head):")
    print(train_data.head())
    
    print("\nTest data (head):")
    print(test_data.head())
    
    # Identify columns
    categorical_cols = train_data.select_dtypes(include=['object', 'category']).columns
    numerical_cols = train_data.select_dtypes(include=['int64', 'float64']).columns

    print(f"\nCategorical columns: {list(categorical_cols)}")
    print(f"Numerical columns: {list(numerical_cols)}")
    
    # Check column alignment
    train_features = train_data.drop(columns=['Sales'], errors='ignore')
    test_features = test_data.drop(columns=['Id'], errors='ignore')
    
    print("\nTraining feature columns:")
    print(train_features.columns)
    
    print("\nTest feature columns:")
    print(test_features.columns)
    
    # Check if columns match
    missing_in_test = set(train_features.columns) - set(test_features.columns)
    missing_in_train = set(test_features.columns) - set(train_features.columns)
    
    if missing_in_test:
        print(f"\nColumns in training but not in test: {missing_in_test}")
    if missing_in_train:
        print(f"\nColumns in test but not in training: {missing_in_train}")
    
    print("\n--- Debugging Complete ---")

# Example usage:
debug_preprocessing(train_store, test_store)
```
