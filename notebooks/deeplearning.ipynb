```python
path_1='data/train_store_merged.csv'
path_2='data/test_store_merged.csv'
```

```python
from scripts.preprocessing import extract_data
train_store=extract_data(path_1)
test_store=extract_data(path_2)
```

```python
feature_train_data = train_store.drop(columns=['Customers', 'Sales'])
target_train_data = train_store['Sales']
feature_test_data = test_store[['Id', 'Store', 'DayOfWeek', 'Open', 'Promo', 'StateHoliday',
       'SchoolHoliday', 'StoreType', 'Assortment', 'CompetitionDistance',
       'CompetitionOpenSinceMonth', 'CompetitionOpenSinceYear', 'Promo2',
       'Promo2SinceWeek', 'Promo2SinceYear', 'PromoInterval', 'Is_holiday',
       'period']]
```

```python
from scripts.deep_learning_mod import create_supervised_data, train_model, scale_data,evaluate_model,build_lstm_model
# Prepare the data
n_lags = 7
X, y = create_supervised_data(target_train_data, n_lags)
X_scaled, y_scaled, scaler = scale_data(X, y)

# Reshape data for LSTM
X_train_lstm = X_scaled[:-1]
y_train_lstm = y_scaled[1:]
X_train_lstm = X_train_lstm.reshape(X_train_lstm.shape[0], X_train_lstm.shape[1], 1)

# Build and train the model
input_shape = (X_train_lstm.shape[1], X_train_lstm.shape[2])
model = build_lstm_model(input_shape)
train_model(model, X_train_lstm, y_train_lstm, epochs=50, batch_size=32)

# Evaluate the model
rmse, y_pred, y_actual = evaluate_model(model, X_train_lstm, y_train_lstm, scaler)
print(f"Root Mean Squared Error: {rmse}")
```
